# Group 51 — Waste Classification Training Notebook

**Models trained:**
- **Primary model** — EfficientNetB0 fine-tuned → 8 waste classes (plastic, paper, metal, glass, organic, e-waste, hazardous, general_trash)
- **Secondary model** — MobileNetV2 fine-tuned → 2 condition classes (clean, contaminated)

**What this notebook does:**
1. Checks GPU and installs all dependencies
2. Mounts Drive and loads the dataset
3. Builds augmented data pipelines
4. Trains primary model in two phases (feature extraction → fine-tuning)
5. Trains secondary condition model
6. Evaluates both models with confusion matrices and classification reports
7. Exports both models to `.h5` and `SavedModel` format
8. Saves everything back to Google Drive

**Runtime:** Make sure you have GPU enabled → Runtime → Change runtime type → T4 GPU

## Cell 4 — Dataset inspection

```
import json
from pathlib import Path
from collections import defaultdict

VALID_EXTENSIONS = {'.jpg', '.jpeg', '.png', '.bmp', '.webp'}

def count_images(split_dir):
    counts = {}
    split_path = Path(split_dir)
    if not split_path.exists():
        return counts
    for cls_dir in sorted(split_path.iterdir()):
        if cls_dir.is_dir():
            n = sum(1 for f in cls_dir.iterdir() if f.suffix.lower() in VALID_EXTENSIONS)
            counts[cls_dir.name] = n
    return counts

train_counts = count_images(TRAIN_DIR)
val_counts   = count_images(VAL_DIR)
test_counts  = count_images(TEST_DIR)

CLASS_NAMES = sorted(train_counts.keys())
NUM_CLASSES = len(CLASS_NAMES)

print(f'Classes found ({NUM_CLASSES}): {CLASS_NAMES}')
print()
print(f'{"Class":<16} {"Train":>8} {"Val":>8} {"Test":>8} {"Total":>9}')
print('─' * 52)
total = 0
for cls in CLASS_NAMES:
    tr = train_counts.get(cls, 0)
    va = val_counts.get(cls, 0)
    te = test_counts.get(cls, 0)
    tot = tr + va + te
    total += tot
    print(f'{cls:<16} {tr:>8,} {va:>8,} {te:>8,} {tot:>9,}')
print('─' * 52)
print(f'{"TOTAL":<16} {sum(train_counts.values()):>8,} {sum(val_counts.values()):>8,} {sum(test_counts.values()):>8,} {total:>9,}')

# Warn if any class is missing
expected = {'plastic','paper','metal','glass','organic','e-waste','hazardous','general_trash'}
missing = expected - set(CLASS_NAMES)
if missing:
    print(f'\nWARNING: These expected classes are missing from the dataset: {missing}')
```

## Cell 5 — Config (all hyperparameters in one place)

```
# ── PRIMARY MODEL CONFIG ───────────────────────────────────────────────────
IMG_SIZE        = 224          # EfficientNetB0 native input size
BATCH_SIZE      = 32           # reduce to 16 if you get OOM errors
EPOCHS_PHASE1   = 10           # feature extraction (frozen base)
EPOCHS_PHASE2   = 20           # fine-tuning (top layers unfrozen)
LR_PHASE1       = 1e-3         # learning rate for phase 1
LR_PHASE2       = 1e-5         # lower LR for fine-tuning
UNFREEZE_FROM   = -30          # unfreeze last 30 layers in phase 2
DROPOUT_RATE    = 0.4
LABEL_SMOOTHING = 0.1          # helps with noisy/contaminated labels

# ── SECONDARY (CONDITION) MODEL CONFIG ────────────────────────────────────
COND_IMG_SIZE     = 224
COND_BATCH_SIZE   = 32
COND_EPOCHS       = 15
COND_LR           = 1e-4
CONDITION_CLASSES = ['clean', 'contaminated']   # binary

# ── PATHS ─────────────────────────────────────────────────────────────────
PRIMARY_MODEL_PATH   = os.path.join(MODELS_DIR, 'primary_model.h5')
SECONDARY_MODEL_PATH = os.path.join(MODELS_DIR, 'condition_model.h5')
PRIMARY_SAVED_PATH   = os.path.join(MODELS_DIR, 'primary_savedmodel')
SECONDARY_SAVED_PATH = os.path.join(MODELS_DIR, 'condition_savedmodel')

# ── REPRODUCIBILITY ────────────────────────────────────────────────────────
SEED = 42
import random, numpy as np
import tensorflow as tf
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print('Config loaded.')
print(f'  Image size   : {IMG_SIZE}x{IMG_SIZE}')
print(f'  Batch size   : {BATCH_SIZE}')
print(f'  Num classes  : {NUM_CLASSES} → {CLASS_NAMES}')
print(f'  Phase 1      : {EPOCHS_PHASE1} epochs @ lr={LR_PHASE1}')
print(f'  Phase 2      : {EPOCHS_PHASE2} epochs @ lr={LR_PHASE2}')
```

## Cell 7 — Build data pipelines (with augmentation)

```
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# EfficientNet expects pixel values in [0, 255] — preprocess_input handles normalisation
from tensorflow.keras.applications.efficientnet import preprocess_input as eff_preprocess

# ── TRAINING AUGMENTATION ─────────────────────────────────────────────────
# Strong augmentation for robustness to real-world conditions (blur, angle, lighting)
train_datagen = ImageDataGenerator(
    preprocessing_function=eff_preprocess,
    rotation_range=30,
    width_shift_range=0.15,
    height_shift_range=0.15,
    shear_range=0.1,
    zoom_range=0.2,
    horizontal_flip=True,
    vertical_flip=False,
    brightness_range=[0.7, 1.3],
    fill_mode='nearest',
)

# ── VALIDATION / TEST — no augmentation, only preprocessing ───────────────
val_datagen = ImageDataGenerator(preprocessing_function=eff_preprocess)

# ── GENERATORS ────────────────────────────────────────────────────────────
train_gen = train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=True,
    seed=SEED,
)

val_gen = val_datagen.flow_from_directory(
    VAL_DIR,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False,
)

test_gen = val_datagen.flow_from_directory(
    TEST_DIR,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False,
)

# Save class index mapping
CLASS_INDICES = train_gen.class_indices
IDX_TO_CLASS  = {v: k for k, v in CLASS_INDICES.items()}
print(f'Class indices: {CLASS_INDICES}')

with open(os.path.join(MODELS_DIR, 'class_indices.json'), 'w') as f:
    json.dump(CLASS_INDICES, f, indent=2)
print('class_indices.json saved to Drive.')
```

## Cell 8 — Compute class weights (handles imbalanced classes like hazardous)

```
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# Get all training labels from generator
train_labels = train_gen.classes

class_weights_array = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_labels),
    y=train_labels
)
CLASS_WEIGHTS = dict(enumerate(class_weights_array))

print('Class weights (higher = rarer class, gets more attention during training):')
for idx, weight in CLASS_WEIGHTS.items():
    print(f'  {IDX_TO_CLASS[idx]:<16} weight = {weight:.3f}')
```

## Cell 9 — Build primary model (EfficientNetB0)

```
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.applications import EfficientNetB0

def build_primary_model(num_classes, img_size=224, dropout_rate=0.4):
    # Load pretrained base — include_top=False removes ImageNet classifier head
    base_model = EfficientNetB0(
        weights='imagenet',
        include_top=False,
        input_shape=(img_size, img_size, 3)
    )
    base_model.trainable = False  # freeze for phase 1

    inputs = tf.keras.Input(shape=(img_size, img_size, 3))
    x = base_model(inputs, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(dropout_rate)(x)
    x = layers.Dense(256, activation='relu', kernel_regularizer=regularizers.l2(1e-4))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(dropout_rate / 2)(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)

    model = models.Model(inputs, outputs)
    return model, base_model

primary_model, base_model = build_primary_model(NUM_CLASSES, IMG_SIZE, DROPOUT_RATE)

primary_model.summary()
print(f'\nTotal params       : {primary_model.count_params():,}')
print(f'Trainable params   : {sum([tf.size(w).numpy() for w in primary_model.trainable_weights]):,}')
print(f'Non-trainable params: {sum([tf.size(w).numpy() for w in primary_model.non_trainable_weights]):,}')
```

## Cell 10 — Define callbacks (used in both training phases)

```
from tensorflow.keras.callbacks import (
    ModelCheckpoint, EarlyStopping, ReduceLROnPlateau, TensorBoard
)
import datetime

LOG_DIR = os.path.join(MODELS_DIR, 'logs', datetime.datetime.now().strftime('%Y%m%d-%H%M%S'))

def get_callbacks(model_path, patience_early=7, patience_lr=3, phase=1):
    return [
        ModelCheckpoint(
            filepath=model_path,
            monitor='val_accuracy',
            save_best_only=True,
            save_weights_only=False,
            verbose=1,
        ),
        EarlyStopping(
            monitor='val_accuracy',
            patience=patience_early,
            restore_best_weights=True,
            verbose=1,
        ),
        ReduceLROnPlateau(
            monitor='val_loss',
            factor=0.5,
            patience=patience_lr,
            min_lr=1e-7,
            verbose=1,
        ),
        TensorBoard(log_dir=os.path.join(LOG_DIR, f'phase{phase}'), histogram_freq=1),
    ]

print('Callbacks defined.')
```

## Cell 11 — Phase 1: Feature extraction (base frozen, train classifier head only)

```
primary_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=LR_PHASE1),
    loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=LABEL_SMOOTHING),
    metrics=['accuracy', tf.keras.metrics.TopKCategoricalAccuracy(k=2, name='top2_accuracy')]
)

print(f'Phase 1: Training classifier head for {EPOCHS_PHASE1} epochs (base frozen)...')
print(f'Trainable layers: {len([l for l in primary_model.layers if l.trainable])} / {len(primary_model.layers)}')

history_phase1 = primary_model.fit(
    train_gen,
    epochs=EPOCHS_PHASE1,
    validation_data=val_gen,
    class_weight=CLASS_WEIGHTS,
    callbacks=get_callbacks(PRIMARY_MODEL_PATH, phase=1),
    verbose=1,
)

print(f'\nPhase 1 complete.')
print(f'Best val accuracy: {max(history_phase1.history["val_accuracy"]):.4f}')
```

## Cell 12 — Phase 2: Fine-tuning (unfreeze top layers of base)

```
# Unfreeze the top layers of the base model
base_model.trainable = True

# Freeze all layers except the last UNFREEZE_FROM layers
for layer in base_model.layers[:UNFREEZE_FROM]:
    layer.trainable = False

trainable_count = sum(1 for l in base_model.layers if l.trainable)
print(f'Unfroze last {UNFREEZE_FROM} layers of base model.')
print(f'Trainable base layers: {trainable_count} / {len(base_model.layers)}')

# Recompile with much lower learning rate
primary_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=LR_PHASE2),
    loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=LABEL_SMOOTHING),
    metrics=['accuracy', tf.keras.metrics.TopKCategoricalAccuracy(k=2, name='top2_accuracy')]
)

print(f'\nPhase 2: Fine-tuning for {EPOCHS_PHASE2} epochs (lr={LR_PHASE2})...')

history_phase2 = primary_model.fit(
    train_gen,
    epochs=EPOCHS_PHASE2,
    validation_data=val_gen,
    class_weight=CLASS_WEIGHTS,
    callbacks=get_callbacks(PRIMARY_MODEL_PATH, patience_early=8, patience_lr=4, phase=2),
    verbose=1,
)

print(f'\nPhase 2 complete.')
print(f'Best val accuracy: {max(history_phase2.history["val_accuracy"]):.4f}')
```

## Cell 13 — Plot training history

```
import matplotlib.pyplot as plt

def combine_histories(h1, h2):
    combined = {}
    for key in h1.history:
        combined[key] = h1.history[key] + h2.history.get(key, [])
    return combined

history_all = combine_histories(history_phase1, history_phase2)
phase1_end  = len(history_phase1.history['accuracy'])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Accuracy
axes[0].plot(history_all['accuracy'],     label='Train accuracy', linewidth=2)
axes[0].plot(history_all['val_accuracy'], label='Val accuracy',   linewidth=2)
axes[0].axvline(x=phase1_end - 0.5, color='gray', linestyle='--', alpha=0.7, label='Phase 1 → 2')
axes[0].set_title('Accuracy over epochs')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Accuracy')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Loss
axes[1].plot(history_all['loss'],     label='Train loss', linewidth=2)
axes[1].plot(history_all['val_loss'], label='Val loss',   linewidth=2)
axes[1].axvline(x=phase1_end - 0.5, color='gray', linestyle='--', alpha=0.7, label='Phase 1 → 2')
axes[1].set_title('Loss over epochs')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Loss')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.suptitle('Primary model — training history', fontsize=13)
plt.tight_layout()
plt.savefig(os.path.join(MODELS_DIR, 'training_history.png'), dpi=120, bbox_inches='tight')
plt.show()
print('Saved training_history.png to Drive.')
```

## Cell 14 — Evaluate primary model on test set

```
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# Load best saved weights
primary_model.load_weights(PRIMARY_MODEL_PATH)

print('Evaluating on test set...')
test_gen.reset()
test_loss, test_acc, test_top2 = primary_model.evaluate(test_gen, verbose=1)
print(f'\nTest accuracy : {test_acc:.4f} ({test_acc*100:.2f}%)')
print(f'Test top-2 acc: {test_top2:.4f} ({test_top2*100:.2f}%)')
print(f'Test loss     : {test_loss:.4f}')

# Per-class predictions
test_gen.reset()
y_pred_probs = primary_model.predict(test_gen, verbose=1)
y_pred       = np.argmax(y_pred_probs, axis=1)
y_true       = test_gen.classes
class_labels = [IDX_TO_CLASS[i] for i in range(NUM_CLASSES)]

print('\nClassification Report:')
print(classification_report(y_true, y_pred, target_names=class_labels, digits=4))

# Save report to Drive
report_text = classification_report(y_true, y_pred, target_names=class_labels, digits=4)
with open(os.path.join(MODELS_DIR, 'classification_report.txt'), 'w') as f:
    f.write(f'Test accuracy : {test_acc:.4f}\n')
    f.write(f'Test top-2 acc: {test_top2:.4f}\n\n')
    f.write(report_text)
print('Classification report saved to Drive.')
```

## Cell 15 — Confusion matrix

```
cm = confusion_matrix(y_true, y_pred)
cm_normalized = cm.astype('float') / cm.sum(axis=1, keepdims=True)

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# Raw counts
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_labels, yticklabels=class_labels, ax=axes[0])
axes[0].set_title('Confusion matrix — raw counts')
axes[0].set_ylabel('True label')
axes[0].set_xlabel('Predicted label')
axes[0].tick_params(axis='x', rotation=45)

# Normalised
sns.heatmap(cm_normalized, annot=True, fmt='.2f', cmap='Blues',
            xticklabels=class_labels, yticklabels=class_labels, ax=axes[1])
axes[1].set_title('Confusion matrix — normalised')
axes[1].set_ylabel('True label')
axes[1].set_xlabel('Predicted label')
axes[1].tick_params(axis='x', rotation=45)

plt.suptitle('Primary model — test set confusion matrix', fontsize=13)
plt.tight_layout()
plt.savefig(os.path.join(MODELS_DIR, 'confusion_matrix.png'), dpi=120, bbox_inches='tight')
plt.show()
print('Saved confusion_matrix.png to Drive.')
```

## Cell 17 — Secondary model: condition classifier (clean vs contaminated)

This model uses the **same images** but relabels them binary: RealWaste and TACO images are treated as 'contaminated', TrashNet/garbage_v2 as 'clean'. The split must already exist in your dataset — if it doesn't, the generator will fall back to whatever subfolders exist under `condition_train/`.

**If you don't have a condition split yet:** skip to Cell 22 and come back after annotating a binary clean/contaminated label on a subset of your images.

```
COND_TRAIN_DIR = os.path.join(LOCAL_DATASET, 'condition_train')
COND_VAL_DIR   = os.path.join(LOCAL_DATASET, 'condition_val')
COND_TEST_DIR  = os.path.join(LOCAL_DATASET, 'condition_test')

has_condition_data = all(os.path.exists(d) for d in [COND_TRAIN_DIR, COND_VAL_DIR, COND_TEST_DIR])

if not has_condition_data:
    print('WARNING: condition split folders not found.')
    print('Expected:')
    print(f'  {COND_TRAIN_DIR}/clean/')
    print(f'  {COND_TRAIN_DIR}/contaminated/')
    print('  (same for condition_val/ and condition_test/)')
    print()
    print('Skipping secondary model training. See Cell 17 note above.')
else:
    print('Condition split found. Proceeding with secondary model training.')
    for d in [COND_TRAIN_DIR, COND_VAL_DIR, COND_TEST_DIR]:
        split = os.path.basename(d)
        for cls in ['clean', 'contaminated']:
            n = len(list(Path(d).joinpath(cls).glob('*'))) if Path(d).joinpath(cls).exists() else 0
            print(f'  {split}/{cls}: {n:,} images')
```

```
if has_condition_data:
    from tensorflow.keras.applications import MobileNetV2
    from tensorflow.keras.applications.mobilenet_v2 import preprocess_input as mob_preprocess

    cond_train_datagen = ImageDataGenerator(
        preprocessing_function=mob_preprocess,
        rotation_range=25,
        width_shift_range=0.1,
        height_shift_range=0.1,
        zoom_range=0.15,
        horizontal_flip=True,
        brightness_range=[0.7, 1.3],
        fill_mode='nearest',
    )
    cond_val_datagen = ImageDataGenerator(preprocessing_function=mob_preprocess)

    cond_train_gen = cond_train_datagen.flow_from_directory(
        COND_TRAIN_DIR, target_size=(COND_IMG_SIZE, COND_IMG_SIZE),
        batch_size=COND_BATCH_SIZE, class_mode='binary', shuffle=True, seed=SEED
    )
    cond_val_gen = cond_val_datagen.flow_from_directory(
        COND_VAL_DIR, target_size=(COND_IMG_SIZE, COND_IMG_SIZE),
        batch_size=COND_BATCH_SIZE, class_mode='binary', shuffle=False
    )
    cond_test_gen = cond_val_datagen.flow_from_directory(
        COND_TEST_DIR, target_size=(COND_IMG_SIZE, COND_IMG_SIZE),
        batch_size=COND_BATCH_SIZE, class_mode='binary', shuffle=False
    )

    # Build MobileNetV2 condition model
    cond_base = MobileNetV2(weights='imagenet', include_top=False,
                            input_shape=(COND_IMG_SIZE, COND_IMG_SIZE, 3))
    cond_base.trainable = False

    cond_inputs = tf.keras.Input(shape=(COND_IMG_SIZE, COND_IMG_SIZE, 3))
    cx = cond_base(cond_inputs, training=False)
    cx = layers.GlobalAveragePooling2D()(cx)
    cx = layers.BatchNormalization()(cx)
    cx = layers.Dropout(0.3)(cx)
    cx = layers.Dense(128, activation='relu')(cx)
    cx = layers.Dropout(0.2)(cx)
    cond_outputs = layers.Dense(1, activation='sigmoid')(cx)   # binary

    condition_model = models.Model(cond_inputs, cond_outputs)
    condition_model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=COND_LR),
        loss='binary_crossentropy',
        metrics=['accuracy', tf.keras.metrics.AUC(name='auc')]
    )

    print('Condition model built. Training...')
    cond_history = condition_model.fit(
        cond_train_gen,
        epochs=COND_EPOCHS,
        validation_data=cond_val_gen,
        callbacks=get_callbacks(SECONDARY_MODEL_PATH, patience_early=5, patience_lr=3),
        verbose=1,
    )

    # Evaluate
    cond_test_gen.reset()
    cond_loss, cond_acc, cond_auc = condition_model.evaluate(cond_test_gen, verbose=1)
    print(f'\nCondition model test accuracy: {cond_acc:.4f}')
    print(f'Condition model test AUC     : {cond_auc:.4f}')

    # Save
    condition_model.save(SECONDARY_SAVED_PATH)
    print(f'Condition model saved → {SECONDARY_SAVED_PATH}')

    # Save condition class indices
    cond_class_indices = cond_train_gen.class_indices
    with open(os.path.join(MODELS_DIR, 'condition_class_indices.json'), 'w') as f:
        json.dump(cond_class_indices, f, indent=2)
    print(f'Condition class indices: {cond_class_indices}')
else:
    print('Skipped — no condition data found.')
```

## Cell 18 — Confidence-aware inference function (primary model)

```
from tensorflow.keras.applications.efficientnet import preprocess_input as eff_preprocess
import numpy as np

# Thresholds from your proposal
HIGH_CONFIDENCE   = 0.80   # trust the result
MEDIUM_CONFIDENCE = 0.55   # show result with caution warning
# below MEDIUM_CONFIDENCE → 'uncertain, please retake photo'

HAZARDOUS_CONFIDENCE_BOOST = 0.10   # lower threshold for hazardous class

def predict_waste(image_path, primary_model, idx_to_class,
                  condition_model=None, cond_idx_to_class=None):
    """
    Run confidence-aware inference on a single image.
    Returns a dict with class, confidence, condition, and guidance tier.
    """
    # Load and preprocess
    try:
        img = Image.open(image_path).convert('RGB').resize((IMG_SIZE, IMG_SIZE))
    except Exception as e:
        return {'error': f'Could not load image: {e}'}

    img_array = np.array(img, dtype=np.float32)
    img_array = eff_preprocess(img_array)
    img_batch  = np.expand_dims(img_array, axis=0)

    # Primary prediction
    probs      = primary_model.predict(img_batch, verbose=0)[0]
    pred_idx   = int(np.argmax(probs))
    pred_class = idx_to_class[pred_idx]
    confidence = float(probs[pred_idx])

    # Top-3 predictions
    top3_idx  = np.argsort(probs)[::-1][:3]
    top3      = [(idx_to_class[i], float(probs[i])) for i in top3_idx]

    # Confidence tier — lower threshold for hazardous
    effective_high = HIGH_CONFIDENCE - (HAZARDOUS_CONFIDENCE_BOOST if pred_class == 'hazardous' else 0)
    effective_med  = MEDIUM_CONFIDENCE - (HAZARDOUS_CONFIDENCE_BOOST if pred_class == 'hazardous' else 0)

    if confidence >= effective_high:
        confidence_tier = 'high'
        guidance_prefix = ''
    elif confidence >= effective_med:
        confidence_tier = 'medium'
        guidance_prefix = 'Likely '
    else:
        confidence_tier = 'low'
        guidance_prefix = 'Uncertain — '

    # Condition prediction (if model available)
    condition = None
    cond_confidence = None
    if condition_model is not None:
        from tensorflow.keras.applications.mobilenet_v2 import preprocess_input as mob_preprocess
        cond_img = Image.open(image_path).convert('RGB').resize((COND_IMG_SIZE, COND_IMG_SIZE))
        cond_arr = mob_preprocess(np.array(cond_img, dtype=np.float32))
        cond_batch = np.expand_dims(cond_arr, axis=0)
        cond_prob   = float(condition_model.predict(cond_batch, verbose=0)[0][0])
        # sigmoid output: 1 = class index 1 in class_indices
        # which class is index 1 depends on cond_idx_to_class
        cond_idx    = 1 if cond_prob >= 0.5 else 0
        condition   = cond_idx_to_class.get(cond_idx, 'unknown') if cond_idx_to_class else ('contaminated' if cond_prob >= 0.5 else 'clean')
        cond_confidence = cond_prob if cond_idx == 1 else (1 - cond_prob)

    return {
        'predicted_class'  : pred_class,
        'confidence'       : round(confidence, 4),
        'confidence_tier'  : confidence_tier,
        'guidance_prefix'  : guidance_prefix,
        'top3'             : top3,
        'condition'        : condition,
        'condition_confidence': round(cond_confidence, 4) if cond_confidence is not None else None,
    }

print('predict_waste() function defined.')
print('Usage: result = predict_waste("path/to/image.jpg", primary_model, IDX_TO_CLASS)')
```

## Cell 19 — Test inference on sample test images

```
import random
from pathlib import Path

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.flatten()

cond_model_for_test = condition_model if has_condition_data else None
cond_idx_map = ({v: k for k, v in cond_train_gen.class_indices.items()}
                if has_condition_data else None)

for i, cls in enumerate(CLASS_NAMES):
    cls_dir = Path(TEST_DIR) / cls
    images  = list(cls_dir.glob('*'))
    images  = [p for p in images if p.suffix.lower() in VALID_EXTENSIONS]
    if not images:
        axes[i].set_title(f'{cls}\n(no test images)', fontsize=8)
        axes[i].axis('off')
        continue

    sample = random.choice(images)
    result = predict_waste(str(sample), primary_model, IDX_TO_CLASS,
                           cond_model_for_test, cond_idx_map)

    try:
        img = Image.open(sample).convert('RGB').resize((224, 224))
        axes[i].imshow(img)
    except Exception:
        axes[i].axis('off')
        continue

    color = 'green' if result['predicted_class'] == cls else 'red'
    title = (
        f"True: {cls}\n"
        f"Pred: {result['predicted_class']} ({result['confidence']:.0%})\n"
        f"[{result['confidence_tier']}]"
    )
    if result['condition']:
        title += f" | {result['condition']}"
    axes[i].set_title(title, fontsize=8, color=color)
    axes[i].axis('off')

plt.suptitle('Sample test predictions (green = correct, red = wrong)', fontsize=12)
plt.tight_layout()
plt.savefig(os.path.join(MODELS_DIR, 'sample_predictions.png'), dpi=120, bbox_inches='tight')
plt.show()
print('Saved sample_predictions.png to Drive.')
```
